#Introduction to TensorFlow

# 📘 TensorFlow Basics

TensorFlow is a powerful open-source library developed by Google for deep learning and numerical computation. It uses dataflow graphs and automatic differentiation to perform optimized computations.

Let's get started by learning how to use Tensors, which are the fundamental building blocks of TensorFlow.


##Installing and Importing TensorFlow

```python
#!pip install tensorflow -q
import tensorflow as tf
print("TensorFlow version:", tf.__version__)
```

## 📦 Tensors

Tensors are multi-dimensional arrays (like NumPy arrays). TensorFlow supports scalar (0D), vector (1D), matrix (2D), and higher dimensional tensors.

Let's see how to create and inspect them.


##Tensor Creation and Attributes

```python
# Scalar
scalar = tf.constant(7)
print("Scalar:", scalar)

# Vector
vector = tf.constant([1.0, 2.0, 3.0])
print("Vector:", vector)

# Matrix
matrix = tf.constant([[1, 2], [3, 4]])
print("Matrix:\n", matrix)

# Check shape and dtype
print("Shape:", matrix.shape)
print("Data Type:", matrix.dtype)
```

## ➕ Tensor Operations

TensorFlow supports element-wise operations like addition, multiplication, matrix multiplication, etc.


##Tensor Arithmetic

```python
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

# Addition
print("Addition:\n", tf.add(a, b))

# Multiplication (element-wise)
print("Element-wise Multiplication:\n", tf.multiply(a, b))

# Matrix multiplication
print("Matrix Multiplication:\n", tf.matmul(a, b))
```

# 🏗️ Building a Neural Network with TensorFlow (Keras)

We'll use `tf.keras`, a high-level API for building and training models. Let's train a simple neural network to classify digits using the MNIST dataset.


##Load Dataset and Normalize

```python
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize to [0, 1]
x_train, x_test = x_train / 255.0, x_test / 255.0
```

```python
print(x_train.shape)
```

##Define a Model


```python
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),  # Input layer
    tf.keras.layers.Dense(128, activation='relu'),  # Hidden layer
    tf.keras.layers.Dense(10, activation='softmax') # Output layer
])
```

##Compile and Train

```python
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=5)
```

##Evaluate the Model

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print("\nTest accuracy :", test_acc)
print("\nTest loss :", test_loss)
```

# 🏗️ Building a Neural Network with PyTorch

Now let's build a simple neural network to classify digits from the MNIST dataset using `torch.nn` and `torch.optim`.


##Load and Normalize MNIST Dataset

```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Transform: Normalize to [0, 1]
transform = transforms.ToTensor()

train_data = datasets.MNIST(root='data', train=True, download=True, transform=transform)
test_data = datasets.MNIST(root='data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)
```

##Define Neural Network

```python
import torch.nn as nn
import torch.nn.functional as F

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(28 * 28, 128)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):
        x = x.view(x.shape[0], -1)  # Flatten
        x = F.relu(self.fc1(x))
        return F.log_softmax(self.fc2(x), dim=1)

model = SimpleNN()
```

##Define Loss and Optimizer

```python
import torch.optim as optim

criterion = nn.NLLLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

##Train the Model

```python
epochs = 3
for epoch in range(epochs):
    for images, labels in train_loader:
        optimizer.zero_grad()
        output = model(images)
        loss = criterion(output, labels)
        loss.backward()
        optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

##Evaluate Accuracy

```python
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        outputs = model(images)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Test Accuracy: {correct / total * 100:.2f}%")
```

# PyTorch vs TensorFlow: Quick Comparison

| Feature                | PyTorch                               | TensorFlow                           |
|------------------------|----------------------------------------|--------------------------------------|
| Execution Mode         | Eager (Dynamic)                        | Eager (Static with option for Graph) |
| High-Level API         | `torch.nn`, `torch.optim`              | `tf.keras`                           |
| Dataset Loader         | `DataLoader`                           | `tf.data.Dataset`                    |
| Training Loop Style    | Mostly manual                          | `model.fit()` or manual              |
| Model Saving           | `torch.save()`                         | `model.save()` or `SavedModel`       |
| Visualization Tool     | TensorBoard (via `torch.utils.tensorboard`) | TensorBoard                    |
| Autograd Support       | Built-in via `autograd`                | `tf.GradientTape()`                  |

Both libraries are capable of building state-of-the-art models. Your choice depends on project needs and comfort with syntax.

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
print("Is GPU available?", tf.config.list_physical_devices('GPU'))
```

```python
# Load and preprocess MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(x_train.shape, y_train.shape, x_test.shape, y_test.shape)
```

```python
# Build the model — exact architecture from the table above
tf_model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),   # Input layer: 28x28 -> 784
    tf.keras.layers.Dense(256, activation='relu'),   # Dense Layer 1
    tf.keras.layers.Dropout(0.3),                    # Dropout 30%
    tf.keras.layers.Dense(128, activation='tanh'),   # Dense Layer 2
    tf.keras.layers.Dense(10, activation='softmax')  # Output layer (Softmax)
])

tf_model.summary()
```

```python
# Evaluate on the MNIST test set
tf_test_loss, tf_test_accuracy = tf_model.evaluate(x_test, y_test, verbose=0)
print(f"TensorFlow Test Accuracy: {tf_test_accuracy:.4f}")
print(f"TensorFlow Test Loss: {tf_test_loss:.4f}")
```

## Part 2 — PyTorch (`torch.nn.Module`)

```python
import time
import torch
import torch.nn as nn
import torch.nn.functional as F
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

print("PyTorch version:", torch.__version__)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
# Load MNIST (same dataset, PyTorch-native loader)
transform = transforms.ToTensor()

pt_train_data = datasets.MNIST(root='data', train=True, download=True, transform=transform)
pt_test_data = datasets.MNIST(root='data', train=False, download=True, transform=transform)

pt_train_loader = DataLoader(pt_train_data, batch_size=64, shuffle=True)
pt_test_loader = DataLoader(pt_test_data, batch_size=64, shuffle=False)
```

```python
# Same architecture as the TensorFlow model above
class SimpleModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()                # Input layer: 28x28 -> 784
        self.fc1 = nn.Linear(784, 256)              # Dense Layer 1
        self.dropout = nn.Dropout(0.3)              # Dropout 30%
        self.fc2 = nn.Linear(256, 128)               # Dense Layer 2
        self.out = nn.Linear(128, 10)                # Output layer

    def forward(self, x):
        x = self.flatten(x)
        x = F.relu(self.fc1(x))                      # ReLU activation
        x = self.dropout(x)
        x = torch.tanh(self.fc2(x))                  # Tanh activation
        x = self.out(x)
        return F.log_softmax(x, dim=1)               # LogSoftmax output

pt_model = SimpleModel().to(device)
print(pt_model)
```

```python
# Adam optimizer + NLLLoss (pairs with LogSoftmax output, equivalent to
# categorical cross-entropy on the softmax probabilities)
optimizer = torch.optim.Adam(pt_model.parameters(), lr=0.001)
criterion = nn.NLLLoss()

start_time = time.time()
for epoch in range(5):
    pt_model.train()
    running_loss, num_batches = 0.0, 0
    for images, labels in pt_train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        output = pt_model(images)
        loss = criterion(output, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        num_batches += 1

    print(f"Epoch {epoch+1}/5 — avg loss: {running_loss/num_batches:.4f}")

print(f"\nPyTorch training time: {time.time() - start_time:.2f} sec")
```

```python
# Evaluate accuracy on the MNIST test set
pt_model.eval()
correct, total = 0, 0

with torch.no_grad():
    for images, labels in pt_test_loader:
        images, labels = images.to(device), labels.to(device)
        output = pt_model(images)
        predictions = output.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

pt_test_accuracy = correct / total
print(f"PyTorch Test Accuracy: {pt_test_accuracy:.4f}")
```
